```python
all_df = []

for i in range(1, 541):
    response = requests.get(
        f'https://api.themoviedb.org/3/movie/top_rated?api_key=9b0f19de78193ecb444a55571e26a12f&page={i}'
    )
    
    data = response.json()
    
    if 'results' in data:
        temp_df = pd.DataFrame(data['results'])[
            ['id','title','overview','popularity','release_date','vote_average','vote_count']
        ]
        all_df.append(temp_df)

df = pd.concat(all_df, ignore_index=True)
```
