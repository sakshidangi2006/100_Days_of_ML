## 2. Opening a csv file from an URL

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Max OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}

req = requests.get(url)

data = StringIO(req.text)
df = pd.read_csv(data)
df.head(2)
```

## 6. use_col parameter
Jab bahut se columns me se kuch hi columns use karne hote hai usecols ka use karte hai

```python
data = pd.read_csv('Countries.csv', usecols=['country','capital_city'
                                     ,'population','title','currency'])
data.head(2)
```

## 7. Squeeze Parameters
Jab ek hi column use karna hota hai to wo dataframe series ban jata hai

```python
s = pd.read_csv('Countries.csv',usecols=['country'])['country']
```
